### Part 2.1: Data collection by Webscraping

```python
# Part 2.1: Use Webscraping to get data.

url2 = "https://en.wikipedia.beta.wmflabs.org/wiki/List_of_best-selling_manga"
reqs2 = requests.get(url2)

reqs2
# Output for reqs2 was <Response [200]>

soup2 = BeautifulSoup(reqs2.text, 'html.parser')


table2 = pd.read_html('https://en.wikipedia.beta.wmflabs.org/wiki/List_of_best-selling_manga')
len(table2)
table2[1]

# Note: table2[1] is the side-bar data we are working with.
# one_hun_mil_sold_df is the list of manga which have sold over 100 million copies.

one_hun_mil_sold_df = table2[1]

# Note: There are square brackets [] found with the values under both 'Approximate sales' and 'Average sales per volume'. 
# The 'Average sales per volume[b]' label also has the [b] bracket.
# These brackets [] are on the wikipedia page because they are used to refer to references.
# However, the square brackets [] need to be removed as they will become obstacles if we want to make use of the label and values. 

# We could use the replace function: replace(r'\[.*?\]', '', regex=True).
# r denotes a raw string. It will have \[ and \] not be treated as escape characters, but as literal backslashes.
# \[ will be a literal [.
# \] will be a literal ].
# Non-greedy Regexes: the .*? part ensures that the regex engine stops capturing characters as soon as it encounters the first closing square bracket ]
# . = Any character (except newline character).
# * = Zero or more occurances.
# ? = Zero or one occurrences.
# [] (non-escaped) = Matches a set of characters inside the bracket. Ex: "[a-m]"
# \ = Signals a special sequence (can be also used to escape special characters)
# regex=True = r'\[.*?\] will be treated as an regular expression (or regex).


# Remove the square brackets [] within the values for 'Approximate sales'.
one_hun_mil_sold_df['Approximate sales'] = one_hun_mil_sold_df['Approximate sales'].replace(r'\[.*?\]', '', regex=True)

# Rename the column 'Average sales per volume[b]' to 'Average sales per volume'.
one_hun_mil_sold_df.rename(columns={'Average sales per volume[b]': 'Average sales per volume'}, inplace=True)

# Remove the square brackets [] within the values for 'Average sales per volume'.
one_hun_mil_sold_df['Average sales per volume'] = one_hun_mil_sold_df['Average sales per volume'].replace(r'\[.*?\]', '', regex=True)

one_hun_mil_sold_df
```

### Part 2.3: Converting Values from Object (strings) to Numerical (from two columns)

```python
# Part 2.3: Convert the Values from Object (strings) to Numerical (from two columns) for Approximate sales and Average sales per volume

import pandas as pd
import re

# The isinstance() function returns True if the specified object is of the specified type, otherwise False.
# Want to find if the value is a string with isinstance()

# re lets you compile, search, match, and manipulate strings using regular expressions.
# re.search() search for a specific pattern in the value string.

# r'(\d+\.?\d*)\s*million'
# \d+ = Matches one or more digits (0-9). Ex: Will match the integer part of a number such as "4" in "4 million".
# \.? = Matches an optional decimal point (.), if the value has a decimal point.
# \d* = Matches zero or more digits after the decimal point, allowing for decimal numbers (ex: "4.6" in "4.6 million").
# \s* = Matches zero or more spaces between the number and the word "million" (to allow for cases where there might be a space).
# million = Matches the literal word "million". value.lower() was used to make the value lower case (Ex: "Million" will become "million").

# match.group(1) = The group(1) method retrieves the first captured group from the regex match. 
# (the part of the pattern inside the parentheses (\d+\.?\d*)), which is the number part of the string.

# Since the value is a string, float(match.group(1)) will make the value within (match.group(1)) become a float. (ex: "4.6" becomes 4.6).  

# Function to convert 'million' text to numeric values
def convert_to_numeric(value):
    if isinstance(value, str):
        match = re.search(r'(\d+\.?\d*)\s*million', value.lower())  # Look for numbers with 'million' as text.
        if match:
            return float(match.group(1)) * 1000000  # Convert and multiply by million and will return a numerical float value.
    return None

# Apply the conversion to the 'Approximate sales' and 'Average sales per volume' columns
one_hun_mil_sold_df['Approximate sales'] = one_hun_mil_sold_df['Approximate sales'].apply(convert_to_numeric)
one_hun_mil_sold_df['Average sales per volume'] = one_hun_mil_sold_df['Average sales per volume'].apply(convert_to_numeric)

# Display the updated dataframe with numeric values in 'Approximate sales' and 'Average sales per volume'
one_hun_mil_sold_df

```

### Part 2.4: Group each demographic, and include for each the sum of approximate sales

```python
# Part 2.4: Group each demographic, then put the sum of approximate sales for each demographic.

# The four demographic groups should be: Shōnen, Seinen, Children and Shōnen/Seinen. 

# Group the dataframe by 'Demographic' and sum the 'Approximate sales' for each group
demographic_sales_sum = one_hun_mil_sold_df.groupby('Demographic')['Approximate sales'].sum().reset_index()

# Display the new dataframe with summed approximate sales for each demographic
print(demographic_sales_sum)
```

### Part 2.5: Plot 2, Seaborn, Data Visualization displaying the sum of approximate sales for each demographic

```python
# Part 2.5: Use Seaborn to plot a bar plot, having the four demographic groups with the sum of the approximate sales.

import seaborn as sns
import matplotlib.pyplot as plt


# Create bar plot, set the bar plot settings.
plt.figure(figsize=(10, 6))
sns.barplot(data=demographic_sales_sum, x='Demographic', y='Approximate sales', hue='Demographic', palette='Set2')

# Have values above each bar.
# enumerate() will generate two values: index and the value.
# Add text to the plot with plt.text
# Inside plt.text(), "index" will be x-coordinate of text value (based off "index" in demographic_sales_sum['Approximate sales'])
# Inside plt.text(), "value" is y-coordinate.
# .0f = Formats value as a number with commas separating thousands (e.g., 2,000,000) and no decimal places.   
for index, value in enumerate(demographic_sales_sum['Approximate sales']):                  
    plt.text(index, value + 5000000, f'{value:,.0f}', ha='center', va='bottom', fontsize=10)

# Adding title, labels, xticks, etc.
plt.title('Demographic with Sum of Approximate Sales',fontsize=14)
plt.xlabel('Demographic',fontsize=12)
plt.ylabel('Sum of Approximate Sales',fontsize=12)
plt.xticks(rotation=45, ha='right')

# Show the plot
plt.tight_layout()
plt.show()

# Answer: Based on the bar plot, it appears that the Shōnen demographic has the highest sum of approximate sales at 2,396,500,000.
# Seinen is at 583,000,000. Children is at 250,000,000. Shōnen/Seinen is at 220,000,000.  
```

### Part 3.1: Combining dataframes of Highest Circulation data with 100 Million Copies data, based on having the same Manga Series

```python
# Part 3.1: Combine Highest Circulation dataframe with the 100 Million Copies dataframe, but only if both dataframes
# have the same Manga series

import pandas as pd


# Find the same Manga series between both dataframes.
# Make a set for the 'Manga series' from both circulation_df and one_hun_mil_sold_df.
# Intersection() will return a set that contains the same items from both circulation_df and one_hun_mil_sold_df.
same_manga_series = set(circulation_df['Manga series']).intersection(set(one_hun_mil_sold_df['Manga series']))

# Filter both dataframes for only the same Manga series.
# Make an updated dataframe (for both Highest Circulation and 100 Million Copies) containing the same manga series from 
# the same_manga_series set using isin() function.
filtered_circulation_df = circulation_df[circulation_df['Manga series'].isin(same_manga_series)]
filtered_one_hun_mil_sold_df = one_hun_mil_sold_df[one_hun_mil_sold_df['Manga series'].isin(same_manga_series)]

# Merge the filtered dataframes on the 'Manga series' column.
# The merge() function is similar to SQL joins and is used when you want to combine DataFrames based on a common column or index. 
# on: Specifies the column(s) to join on.
merged_df = pd.merge(filtered_circulation_df, filtered_one_hun_mil_sold_df, on='Manga series')
merged_df
```

## Part 3.2: Plot 4, Seaborn, Data Visualization with Scatter Plot of Total Circulation vs Approximate Sales

```python
# Part 3.2: Use Seaborn to create Scatter Plot of Total Circulation vs Approximate Sales from each Manga Series. 

import seaborn as sns
import matplotlib.pyplot as plt

# Create scatter plot, set scatter plot settings.
sns.scatterplot(x='Total circulation (est.)', y='Approximate sales', hue='Manga series', data=merged_df, 
                palette='Set2', s=100)  # Could adjust marker size

# Add title and labels
plt.title('Total Circulation vs Approximate Sales',fontsize=14)
plt.xlabel('Total Circulation (est.)',fontsize=12)
plt.ylabel('Approximate Sales',fontsize=12)

# Legend settings
plt.legend(title='Manga Series', bbox_to_anchor=(1.05, 1), loc='upper left')

# Show plot
plt.show()

# Answer: Based on the scatter plot (limited data), it appears that a higher circulation could contribute 
# to higher sales (with the exception of Jojo's Bizarre Adventure and KochiKame).
```

## Part 3.3: Plot 5, Seaborn, Data Visualization with Double Bar Plot of Manga Series vs Number of Issues and Volumes, with a second plot displaying Manga Series vs (Average) Number of Issues per Volume

```python
# Part 3.3: Use Seaborn to create Double Bar Plot of Manga Series vs Number of Issues and Volumes.
# Then make another bar plot next to the Double Bar Plot displaying Manga Series vs (Average) Number of Issues per Volume.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Note: For this question, to save space, I will only plot 3 manga series: One Piece, Naruto and Bleach)

# Filter for the specific Manga series: One Piece, Naruto, and Bleach
# Make an updated dataframe containing only One Piece, Naruto, and Bleach from merged_df using isin() function.
filtered_df = merged_df[merged_df['Manga series'].isin(['One Piece', 'Naruto', 'Bleach'])]


# First plot: Issues vs No. of Collected Volumes
# Use the melt() function: the pd.melt function is used to unpivot 'Issues' and 'No. of collected volumes' (they will become values)
# while 'Manga series' will remain as a identifier. (the rows for One Piece, Naruto, and Bleach will double their rows, one for 'Issues' 
# and the other for 'No. of collected volumes'.
# id_vars = 'Manga series' will remain as a identifier
# var_name = New column label or identifier called 'Form of Literature'.
# value_vars = 'Issues' and 'No. of collected volumes' will now become values under the 'Form of Literature' identifier.
# value_name = New column label or identifier called 'Count'. The original values from 'Issues' and 'No. of collected volumes' will
# be stored here, with 'Issues' and 'No. of collected volumes' displayed horizontally accordingly. 
manga_data = filtered_df[['Manga series', 'Issues', 'No. of collected volumes']]
melted_df = manga_data.melt(id_vars='Manga series', value_vars=['Issues', 'No. of collected volumes'],
                            var_name='Form of Literature', value_name='Count')
#melted_df

# Second plot: Average Issues per Volume
# A new column called 'Average Issues per Volume' will be created.
# The 'Average Issues per Volume' values will be equal to the values of 'Issues' divided by the value of 'No. of collected volumes'.
filtered_df_copy = filtered_df.copy()
filtered_df_copy.loc[:, 'Average Issues per Volume'] = filtered_df_copy['Issues'] / filtered_df_copy['No. of collected volumes']

# Create a figure with two subplots (side by side)
# Will have 1 row of subplots.
# Will have 2 columns of subplots.
fig, axes = plt.subplots(1, 2, figsize=(16, 6))


# Note: With two subplots, be aware to have settings as per example below:
# axes[0].plot(x_data, y_data)  # Plot on the first subplot (left)
# axes[1].scatter(x_data, y_data)  # Plot on the second subplot (right)


# First plot: Manga Series vs Issues and No. of Collected Volumes. Double Bar Graph.
sns.barplot(x='Manga series', y='Count', hue='Form of Literature', data=melted_df, palette='Set2', ax=axes[0])
axes[0].set_title('Manga Series vs Issues and Number of Volumes',fontsize=16)
axes[0].set_xlabel('Manga Series',fontsize=14)
axes[0].set_ylabel('Count',fontsize=14)

# Display the value above each bar for the first plot.
# axes[0] = Represents the first subplot.
# .patches = A patch represents a single bar in the bar plot. 
# axes[0].text() adds text to the plot.
# p.get_x() = This gets the x-coordinate of the left edge of the bar.
# p.get_width() = This gets the width of the bar. (Distance from left edge to right edge of bar).
# (p.get_x() + p.get_width())/2 = horizontal midpoint of the bar.
# p.get_height() = This gets the height of the bar (the value of the bar).
for p in axes[0].patches:
    if p.get_height() > 0:  # Check if the bar has a height greater than 0. This removes a "0" being displayed in between the two bars above One Piece.
        axes[0].text(p.get_x() + p.get_width() / 2., p.get_height() + 10,
                     f'{int(p.get_height())}', ha='center', va='bottom', fontsize=10)

# Second plot: Average Issues per Volume.
sns.barplot(x='Manga series', y='Average Issues per Volume', hue='Manga series', data=filtered_df_copy, palette='Set1', ax=axes[1])
axes[1].set_title('Manga Series vs Average Issues per Volume',fontsize=16)
axes[1].set_xlabel('Manga Series',fontsize=14)
axes[1].set_ylabel('Average Issues per Volume',fontsize=14)

# Display the value above each bar for the second plot.
for p in axes[1].patches:
    axes[1].text(p.get_x() + p.get_width() / 2., p.get_height() + 0.1, 
                 f'{p.get_height():.2f}', ha='center', va='bottom', fontsize=10)

# Adjust layout, and show plot.
plt.tight_layout()
plt.show()

# Note: To my understanding one magazine issue contains one chapter from a volume for a manga series 
# (a magazine issue usually contains a collection of one chapter from different manga series). 
# This problem is to observe how many chapters on average are found in a volume.

# Answer: Based on the bar plots, it appears that the number of issues to number of volumes have a relation close to or above a 9 to 1 ratio.
# (We could expect a volume from one of the manga series will usually have an average of 9 issues (or chapters) or more per volume.)
```
